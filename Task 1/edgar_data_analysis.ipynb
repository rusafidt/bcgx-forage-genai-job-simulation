# EDGAR Financial Analysis (Structured)

This notebook analyzes 10-K financial data for **Apple, Microsoft, and Tesla**.

Following are the steps:
1. Import pandas
2. Load the CSV data
3. Inspect and sort the dataset
4. Calculate year-over-year growth (%)
5. Create simple summary tables
6. Summarizing the findings


## Step 1: Import pandas

**What this does:** Loads pandas so can work with table-like data.


```python
import pandas as pd
```

## Step 2: Load the CSV file

**What this does:** Reads the extracted EDGAR data into a DataFrame named `df`.

**Expected output:** First 5 rows of dataset.


```python
df = pd.read_csv('data/EDGAR Entity Extracted Data.csv')
df.head()
```

## Step 3: Understand and organize the data

**What this does:**
- Checks column names and data types
- Sorts data by `Company` and `Fiscal Year` for trend analysis


```python
df.info()
```

```python
df = df.sort_values(['Company', 'Fiscal Year']).reset_index(drop=True)
df[['Company', 'Fiscal Year', 'Total Revenue', 'Net Income']].head(9)
```

## Step 4: Calculate year-over-year growth (%)

**What this does:** Calculates percentage change from one year to the next for each company.

Formula used by pandas: `pct_change() * 100`


```python
df['Revenue Growth (%)'] = df.groupby('Company')['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby('Company')['Net Income'].pct_change() * 100
df['Total Assets Growth (%)'] = df.groupby('Company')['Total Assets'].pct_change() * 100
df['Total Liabilities Growth (%)'] = df.groupby('Company')['Total Liabilities'].pct_change() * 100
df['Operating Cash Flow Growth (%)'] = df.groupby('Company')['Cash Flow from Operating Activities'].pct_change() * 100
df
```

## Step 5: Summarize trends

**What this does:** Creates simple summaries to compare companies.


```python
growth_cols = [
    'Revenue Growth (%)',
    'Net Income Growth (%)',
    'Total Assets Growth (%)',
    'Total Liabilities Growth (%)',
    'Operating Cash Flow Growth (%)'
]
company_growth_summary = df.groupby('Company')[growth_cols].mean()
company_growth_summary
```

```python
metric_cols = [
    'Total Revenue',
    'Net Income',
    'Total Assets',
    'Total Liabilities',
    'Cash Flow from Operating Activities'
]
yearly_totals = df.groupby('Fiscal Year')[metric_cols].sum()
yearly_totals
```

## Step 6: Summarizing the findings

- Microsoft shows the strongest and most consistent growth. Revenue and net income both increased a lot in 2024 and 2025, and operating cash flow also stayed strong.
- Apple looks more stable and moderate. Revenue grew slowly in 2024, then improved in 2025. Net income dipped in 2024 but recovered in 2025.
- Tesla looks the most volatile. Revenue growth became very weak and then negative in 2025, while net income dropped sharply in both 2024 and 2025.
- For balance sheet trends, Microsoft and Tesla both increased assets and liabilities, but Tesla’s profitability did not keep up. Apple reduced both assets and liabilities in 2025.
- Overall, Microsoft appears financially strongest in this period, Apple is relatively steady, and Tesla shows pressure on earnings despite higher scale.
